### Copy Operation : 

```python
lst1 = [1,2,3,4,5]
lst2 = lst1
lst2
```

```python
lst2[3] = 1000
lst2
```

```python
lst1
```

```python
lst1 = [1,2,3,4,5]
lst2 = lst1.copy() 
```

```python
lst2[2] = 300
lst2
```

```python
lst1,lst2
```

```python
def add_item(item, items=[]):
    items.append(item)
    return items

print(add_item(1))
print(add_item(2))
print(add_item(3))
```
