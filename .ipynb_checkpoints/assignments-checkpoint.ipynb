```python
##### Assg 8 : Asks the user to input numbers until they input 0. The program should print the sum of all the input numbers.
total_sum = 0
while True :
    num = int(input(" Enter a number "))

    if num == 0:
        break
    total_sum += num
    
print(f"The Number you have Enterd is Zero. The Code End's Here!! So the sum is {total_sum}")
        
```

```python
# Assg 9 : Print all the numbers from 1 to 10 except 5 using a for loop and continue statement.

for i in range(1, 11):
    if i == 5:
        continue
    print(i)
```

```python
# Ass 11 : Asks the user to input a number and prints all the even numbers from 1 to that number using a for loop.

num = int(input("Enter a Number: "))
for i in range(1, num + 1):
    if i % 2 == 0:
        print(i)
```

```python
# Assg 12 : Write a Factorial Prg
num = int(input(" Enter a number: "))
fact = 1
if num < 0:
    print(" For -ve number there is no Factorial number")
elif num == 0:
    print("Ther entred number is Zero")
else:
    for i in range(1, num+1):
        fact *= i
    print(fact)
         
```
